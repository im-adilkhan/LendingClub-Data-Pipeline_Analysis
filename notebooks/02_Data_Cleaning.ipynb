```python
# Step 1  → Import Libraries
# Step 2  → Load Dataset
# Step 3  → Shape
# Step 4  → Copy Dataset
# Step 5  → Info
# Step 6  → Missing Value Report
# Step 7  → High Missing Analysis
# Step 8  → Review High Missing Columns
# Step 9  → Remove Unnecessary Columns
# Step 10 → Remove Duplicates
# Step 11 → Convert Date Columns
# Step 12 → Handle Missing Values
# Step 13 → Validate Dataset
# ✅ Step 14 → Select Final Business Columns
# Step 15 → Save clean_loan.csv


import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows",100)
```

```python
df = pd.read_csv(
    r"C:\Users\henry\OneDrive\Desktop\LendingClub-Data-Pipeline\data\raw\loan.csv",
    low_memory=False
)
```

```python
print(df.shape)
```

```python

loan_df = df.copy()
```

```python
loan_df.info()
```

```python
missing = pd.DataFrame({
    "missing_value" : loan_df.isnull().sum(),
    "missing %" : (loan_df.isnull().mean()*100).round(2)
})


missing = missing.sort_values(
    by="missing_value",
    ascending=False
)

missing.head(30)
```

```python
high_missing = missing[missing["missing %"] > 80]

high_missing
```

```python
print( 
      "Columns having more than 80% missing values :",
      high_missing.shape[0])
```

```python
high_missing_cols = high_missing.index.tolist()

high_missing_cols
```

```python
loan_df.drop(
    columns=high_missing_cols,
    inplace=True
)
```

```python
loan_df.shape
```

```python
drop_cols = [

    "id",
    "member_id",
    "url",
    "desc",
    "policy_code"

]
```

```python
loan_df.drop(
    columns=drop_cols,
    inplace=True,
    errors="ignore"
)

# errors="ignore"

# Agar column already remove ho gaya ho to error nahi aayega.

loan_df.shape
```

```python
loan_df.duplicated().sum()
```

```python
date_cols = [
    "issue_d",
    "last_pymnt_d",
    "next_pymnt_d",
    "last_credit_pull_d"
]
```

```python
loan_df[date_cols].dtypes
```

```python
numeric_cols = loan_df.select_dtypes(
    include=np.number
).columns

numeric_cols
```

```python
loan_df.isnull().sum().sort_values(ascending=False)
```

```python
num_cols = loan_df.select_dtypes(include=["number"]).columns




for col in num_cols:
    loan_df[col] = loan_df[col].fillna(loan_df[col].median())
    
```

```python
loan_df.isnull().sum().sort_values(ascending=False)
```

```python
cat_cols = loan_df.select_dtypes(include=["object"]).columns

for col in cat_cols:
    loan_df[col] = loan_df[col].fillna("unknown")

```

```python
loan_df.isnull().sum().sort_values(ascending=False)
```

```python
# Loan Information
loan_columns = [
    "loan_amnt",
    "funded_amnt",
    "funded_amnt_inv",
    "term",
    "int_rate",
    "installment",
    "purpose",
    "loan_status"
]

# Customer Information
customer_columns = [
    "emp_title",
    "emp_length",
    "home_ownership",
    "annual_inc",
    "verification_status",
    "application_type",
    "addr_state",
    "zip_code"
]

# Credit Profile
credit_columns = [
    "grade",
    "sub_grade",
    "dti",
    "delinq_2yrs",
    "earliest_cr_line",
    "inq_last_6mths",
    "open_acc",
    "pub_rec",
    "revol_bal",
    "revol_util",
    "total_acc",
    "acc_now_delinq"
]

# Payment Information
payment_columns = [
    "out_prncp",
    "total_pymnt",
    "total_rec_prncp",
    "total_rec_int",
    "recoveries",
    "last_pymnt_d",
    "last_pymnt_amnt",
    "last_credit_pull_d",
    "issue_d",
    "total_rev_hi_lim"
]

# Final Selected Columns
selected_columns = (
    loan_columns +
    customer_columns +
    credit_columns +
    payment_columns
)

# Keep only selected columns
loan_df = loan_df[selected_columns]
```

```python
print("Shape:", loan_df.shape)
print("Columns:", len(loan_df.columns))
print(loan_df.columns.tolist())
```

```python
loan_df.to_csv(
    r"C:\Users\henry\OneDrive\Desktop\LendingClub-Data-Pipeline\data\processed\clean_loan.csv",
    index=False
)


print("Clean dataset saved successfully.")
```
